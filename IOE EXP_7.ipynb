```python

# ============================================
# Smart Stadium Entry Gate Management System
# Dynamic Density-Based Gate Controller
# Google Colab Version
# ============================================

import random
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time

# Gate names
gates = ["Gate A", "Gate B", "Gate C", "Gate D"]

# Initial queue of fans
queues = [10, 6, 15, 8]

# Number of monitoring cycles
simulation_steps = 20

for step in range(1, simulation_steps + 1):

    # Random new fans arrive
    new_fans = [random.randint(0, 8) for _ in range(4)]
    queues = [queues[i] + new_fans[i] for i in range(4)]

    # Find busiest gate
    active_gate = queues.index(max(queues))
    active_gate_name = gates[active_gate]

    # Fans processed
    fans_cleared = min(queues[active_gate], random.randint(6, 12))
    queues[active_gate] -= fans_cleared

    # Clear previous graph
    clear_output(wait=True)

    # Colors
    colors = ['red', 'red', 'red', 'red']
    colors[active_gate] = 'green'

    # Plot graph
    plt.figure(figsize=(8,5))
    plt.bar(gates, queues, color=colors)

    plt.ylim(0,40)
    plt.ylabel("Number of Fans Waiting")
    plt.xlabel("Entry Gates")
    plt.title(f"Monitoring Cycle {step}: Extra Staff -> {active_gate_name}")

    # Display values on bars
    for i, value in enumerate(queues):
        plt.text(i, value+0.5, str(value),
                 ha='center', fontsize=12)

    plt.figtext(
        0.5,
        0.01,
        f"Action: Processed {fans_cleared} fans at {active_gate_name}",
        ha='center',
        fontsize=12,
        bbox=dict(facecolor='white', edgecolor='black')
    )

    plt.show()

    # Wait before next cycle
    time.sleep(1.5)

print("Simulation Completed Successfully!")
```
