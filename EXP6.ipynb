```python
# ======================================================
# Threat Modeling & Security Hardening
# IoE Smart Agriculture System
# Google Colab
# ======================================================

import random
import hashlib
import pandas as pd
import matplotlib.pyplot as plt

print("="*60)
print(" IoE SMART AGRICULTURE SECURITY SYSTEM ")
print("="*60)

# ----------------------------------------------------
# User Authentication
# ----------------------------------------------------

username = "admin"
password = "Farm@123"

password_hash = hashlib.sha256(password.encode()).hexdigest()

print("\nLogin Successful")
print("Username :", username)
print("Password Hash :", password_hash[:20],"...")

# ----------------------------------------------------
# Smart Farm Sensor Records
# ----------------------------------------------------

field_ids = ["FIELD001","FIELD002","FIELD003","FIELD004","FIELD005"]

records = []

for field in field_ids:

    sensor = random.randint(1001,1050)
    moisture = random.randint(20,100)

    threat = random.choice([
        "SAFE",
        "Sensor Spoofing",
        "Unauthorized Access",
        "Data Tampering",
        "DoS Attack"
    ])

    if threat=="SAFE":
        status="SECURE"
    else:
        status="BLOCKED"

    records.append([
        field,
        sensor,
        moisture,
        threat,
        status
    ])

# ----------------------------------------------------
# Display Sensor Records
# ----------------------------------------------------

df = pd.DataFrame(records,columns=[
    "Field ID",
    "Sensor ID",
    "Soil Moisture (%)",
    "Threat",
    "Status"
])

print("\nSmart Agriculture Report\n")
print(df)

# ----------------------------------------------------
# Security Summary
# ----------------------------------------------------

safe = len(df[df["Threat"]=="SAFE"])
spoof = len(df[df["Threat"]=="Sensor Spoofing"])
unauth = len(df[df["Threat"]=="Unauthorized Access"])
tamper = len(df[df["Threat"]=="Data Tampering"])
dos = len(df[df["Threat"]=="DoS Attack"])

print("\n========== SECURITY REPORT ==========")
print("Safe Sensors        :",safe)
print("Sensor Spoofing     :",spoof)
print("Unauthorized Access :",unauth)
print("Data Tampering      :",tamper)
print("DoS Attack          :",dos)

# ----------------------------------------------------
# GRAPH 1
# Soil Moisture
# ----------------------------------------------------

plt.figure(figsize=(8,4))
plt.bar(df["Field ID"],df["Soil Moisture (%)"])
plt.title("Field Soil Moisture")
plt.xlabel("Field")
plt.ylabel("Moisture (%)")
plt.grid(axis="y")
plt.show()

# ----------------------------------------------------
# GRAPH 2
# Secure vs Blocked
# ----------------------------------------------------

secure = len(df[df["Status"]=="SECURE"])
blocked = len(df[df["Status"]=="BLOCKED"])

plt.figure(figsize=(5,4))
plt.bar(["SECURE","BLOCKED"],[secure,blocked])
plt.title("Security Status")
plt.ylabel("Count")
plt.show()

# ----------------------------------------------------
# GRAPH 3
# Threat Distribution
# ----------------------------------------------------

plt.figure(figsize=(6,6))
plt.pie(
    [safe,spoof,unauth,tamper,dos],
    labels=[
        "SAFE",
        "Sensor Spoofing",
        "Unauthorized",
        "Tampering",
        "DoS"
    ],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Threat Distribution")
plt.show()

# ----------------------------------------------------
# GRAPH 4
# Sensor Moisture Readings
# ----------------------------------------------------

plt.figure(figsize=(8,4))
plt.plot(df["Sensor ID"],df["Soil Moisture (%)"],marker="o")
plt.title("Sensor Moisture Readings")
plt.xlabel("Sensor ID")
plt.ylabel("Soil Moisture (%)")
plt.grid(True)
plt.show()

print("\n======================================")
print("Threat Modeling Completed Successfully")
print("Security Hardening Applied")
print("System Protected Successfully")
print("======================================")
```
