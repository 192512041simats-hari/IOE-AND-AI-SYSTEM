```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

# Create sample dataset
data = {
    "ExamDuration(min)": [60, 75, 90, 80, 70, 85, 95, 88, 76, 82],
    "ScreenSwitches": [2, 5, 1, 4, 3, 6, 8, 5, 2, 4],
    "FaceDetections": [100, 98, 100, 97, 99, 96, 95, 97, 100, 99],
    "MouseClicks": [150, 180, 165, 170, 160, 190, 210, 185, 155, 175],
    "SuspiciousScore": [5, 20, 3, 15, 8, 30, 45, 25, 6, 18]
}

df = pd.DataFrame(data)

# Display original data
print("Original Dataset:")
print(df)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Remove duplicates
df = df.drop_duplicates()

# Normalize data
scaler = MinMaxScaler()
df[df.columns] = scaler.fit_transform(df)

print("\nNormalized Dataset:")
print(df)

# Plot Suspicious Score
plt.plot(df["SuspiciousScore"], marker='o')
plt.title("Normalized Suspicious Score")
plt.xlabel("Student Sample")
plt.ylabel("Suspicious Score")
plt.grid(True)
plt.show()
```
