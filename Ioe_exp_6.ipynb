```python

# ======================================================
# Threat Modeling & Security Hardening
# IoE Smart Banking System
# Google Colab
# ======================================================

import random
import hashlib
import pandas as pd
import matplotlib.pyplot as plt

print("="*60)
print(" IoE SMART BANKING SECURITY SYSTEM ")
print("="*60)

# ----------------------------------------------------
# User Authentication
# ----------------------------------------------------

username = "admin"
password = "Bank@123"

password_hash = hashlib.sha256(password.encode()).hexdigest()

print("\nLogin Successful")
print("Username :", username)
print("Password Hash :", password_hash[:20],"...")

# ----------------------------------------------------
# Smart ATM Transactions
# ----------------------------------------------------

atm_ids = ["ATM001","ATM002","ATM003","ATM004","ATM005"]

records = []

for atm in atm_ids:

    customer = random.randint(1001,1050)
    amount = random.randint(500,10000)

    threat = random.choice([
        "SAFE",
        "ATM Spoofing",
        "Brute Force",
        "Data Tampering",
        "DoS Attack"
    ])

    if threat=="SAFE":
        status="SUCCESS"
    else:
        status="BLOCKED"

    records.append([
        atm,
        customer,
        amount,
        threat,
        status
    ])

# ----------------------------------------------------
# Display Transactions
# ----------------------------------------------------

df = pd.DataFrame(records,columns=[
    "ATM ID",
    "Customer ID",
    "Amount",
    "Threat",
    "Status"
])

print("\nATM Transaction Report\n")
print(df)

# ----------------------------------------------------
# Security Summary
# ----------------------------------------------------

safe = len(df[df["Threat"]=="SAFE"])
spoof = len(df[df["Threat"]=="ATM Spoofing"])
brute = len(df[df["Threat"]=="Brute Force"])
tamper = len(df[df["Threat"]=="Data Tampering"])
dos = len(df[df["Threat"]=="DoS Attack"])

print("\n========== SECURITY REPORT ==========")
print("Safe Transactions :",safe)
print("ATM Spoofing      :",spoof)
print("Brute Force       :",brute)
print("Data Tampering    :",tamper)
print("DoS Attack        :",dos)

# ----------------------------------------------------
# GRAPH 1
# Transaction Amount
# ----------------------------------------------------

plt.figure(figsize=(8,4))
plt.bar(df["ATM ID"],df["Amount"])
plt.title("ATM Transaction Amount")
plt.xlabel("ATM")
plt.ylabel("Amount")
plt.grid(axis="y")
plt.show()

# ----------------------------------------------------
# GRAPH 2
# Success vs Blocked
# ----------------------------------------------------

success = len(df[df["Status"]=="SUCCESS"])
blocked = len(df[df["Status"]=="BLOCKED"])

plt.figure(figsize=(5,4))
plt.bar(["SUCCESS","BLOCKED"],[success,blocked])
plt.title("Transaction Status")
plt.ylabel("Count")
plt.show()

# ----------------------------------------------------
# GRAPH 3
# Threat Distribution
# ----------------------------------------------------

plt.figure(figsize=(6,6))
plt.pie(
    [safe,spoof,brute,tamper,dos],
    labels=[
        "SAFE",
        "ATM Spoofing",
        "Brute Force",
        "Tampering",
        "DoS"
    ],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Threat Distribution")
plt.show()

# ----------------------------------------------------
# GRAPH 4
# Customer Transaction Amount
# ----------------------------------------------------

plt.figure(figsize=(8,4))
plt.plot(df["Customer ID"],df["Amount"],marker="o")
plt.title("Customer Transaction Amount")
plt.xlabel("Customer ID")
plt.ylabel("Amount")
plt.grid(True)
plt.show()

print("\n======================================")
print("Threat Modeling Completed Successfully")
print("Security Hardening Applied")
print("System Protected Successfully")
print("======================================")
```
