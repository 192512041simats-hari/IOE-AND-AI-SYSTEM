```python

# ======================================================
# Threat Modeling & Security Hardening
# IoE Smart Home Security Hub
# Google Colab
# ======================================================

import random
import hashlib
import pandas as pd
import matplotlib.pyplot as plt

print("="*60)
print(" IoE SMART HOME SECURITY HUB ")
print("="*60)

# ----------------------------------------------------
# User Authentication
# ----------------------------------------------------

username = "homeowner"
password = "Home@2026"

password_hash = hashlib.sha256(password.encode()).hexdigest()

print("\nLogin Successful")
print("Username :", username)
print("Password Hash :", password_hash[:20],"...")

# ----------------------------------------------------
# Smart Device Access Events
# ----------------------------------------------------

device_ids = ["SmartLock01","SmartCam02","DoorSensor03","Thermostat04","Gateway05"]

records = []

for device in device_ids:

    user_id = random.randint(101,150)
    access_attempts = random.randint(1,20)

    threat = random.choice([
        "SAFE",
        "Credential Stuffing",
        "Firmware Tampering",
        "Replay Attack",
        "DoS Attack"
    ])

    if threat=="SAFE":
        status="SUCCESS"
    else:
        status="BLOCKED"

    records.append([
        device,
        user_id,
        access_attempts,
        threat,
        status
    ])

# ----------------------------------------------------
# Display Access Log
# ----------------------------------------------------

df = pd.DataFrame(records,columns=[
    "Device ID",
    "User ID",
    "Access Attempts",
    "Threat",
    "Status"
])

print("\nDevice Access Report\n")
print(df)

# ----------------------------------------------------
# Security Summary
# ----------------------------------------------------

safe = len(df[df["Threat"]=="SAFE"])
cred = len(df[df["Threat"]=="Credential Stuffing"])
firmware = len(df[df["Threat"]=="Firmware Tampering"])
replay = len(df[df["Threat"]=="Replay Attack"])
dos = len(df[df["Threat"]=="DoS Attack"])

print("\n========== SECURITY REPORT ==========")
print("Safe Access Events    :",safe)
print("Credential Stuffing   :",cred)
print("Firmware Tampering    :",firmware)
print("Replay Attack         :",replay)
print("DoS Attack            :",dos)

# ----------------------------------------------------
# GRAPH 1
# Access Attempts per Device
# ----------------------------------------------------

plt.figure(figsize=(8,4))
plt.bar(df["Device ID"],df["Access Attempts"])
plt.title("Access Attempts per Device")
plt.xlabel("Device")
plt.ylabel("Access Attempts")
plt.grid(axis="y")
plt.show()

# ----------------------------------------------------
# GRAPH 2
# Success vs Blocked
# ----------------------------------------------------

success = len(df[df["Status"]=="SUCCESS"])
blocked = len(df[df["Status"]=="BLOCKED"])

plt.figure(figsize=(5,4))
plt.bar(["SUCCESS","BLOCKED"],[success,blocked])
plt.title("Access Status")
plt.ylabel("Count")
plt.show()

# ----------------------------------------------------
# GRAPH 3
# Threat Distribution
# ----------------------------------------------------

plt.figure(figsize=(6,6))
plt.pie(
    [safe,cred,firmware,replay,dos],
    labels=[
        "SAFE",
        "Credential Stuffing",
        "Firmware Tampering",
        "Replay Attack",
        "DoS"
    ],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Threat Distribution")
plt.show()

# ----------------------------------------------------
# GRAPH 4
# User Access Attempts
# ----------------------------------------------------

plt.figure(figsize=(8,4))
plt.plot(df["User ID"],df["Access Attempts"],marker="o")
plt.title("User Access Attempts")
plt.xlabel("User ID")
plt.ylabel("Access Attempts")
plt.grid(True)
plt.show()

print("\n======================================")
print("Threat Modeling Completed Successfully")
print("Security Hardening Applied")
print("System Protected Successfully")
print("======================================")
```
